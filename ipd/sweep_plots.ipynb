```python
from pathlib import Path

import pandas as pd

from ipd.sweep_plots import (
    plot_sweep_summary,
    suggest_sweep_cols,
)

# Aliases for axis labeling
BASE_DIR = "/path/to/user-data/mkazman/PFM_runs/train_latents/"
_ckpt1 = BASE_DIR + "ame_afdb_omol_partial_cw_256/store/ame_afdb_omol_partial_cw_256/checkpoints/"
_ckpt2 = BASE_DIR + "ame_afdb_mgnify_omol_partial_cw_256/store/ame_afdb_mgnify_omol_partial_cw_256/checkpoints/"
VALUE_LABELS = {
    "ckpt_path": {
        _ckpt1: "afdb_omol",
        _ckpt2: "mgnify_afdb_omol",
    }
}

SWEEP_RUNDIR_PATH = Path("/path/to/user-data/rib7/PFM_runs/inference/sweeps/task_temp_ckpt_2026-03-10")
SWEEP_DF = pd.read_csv(SWEEP_RUNDIR_PATH / "sweep_per_example_metrics.csv")

# Broadcasts diversity to a per-example metric, based on the overall diversity for that task
per_task_df = pd.read_csv(SWEEP_RUNDIR_PATH / "sweep_per_task_metrics.csv")

# Identify the sweep axes
sweep_cols = suggest_sweep_cols(SWEEP_DF, sweep_config_path=(SWEEP_RUNDIR_PATH / "sweep_config.yaml"))
# inspect the printed table, edit the list if needed
```

```python
sweep_cols = [c for c in sweep_cols if c != "ckpt_name"]

plot_sweep_summary(
    SWEEP_DF,
    sweep_cols,
    metric_col=["clustered_pass_rate"],
    heatmap_col1="bb_ca_noise_scale",
    heatmap_col2="latents_noise_scale",
    marginal_sweep_cols=["bb_ca_noise_scale", "latents_noise_scale"],
    task_col="dataset_name",  # Plotting per-dataset: switch to "task" to plot results per-task
    ylabel="pass_rate",
    higher_is_better=True,  # When determining "best values" that will be printed
    value_labels=VALUE_LABELS,
    cols_to_iterate=["ckpt_path"],
)
```

```python
pass_rate_cols = [c for c in SWEEP_DF.columns if c.endswith("pass_rate") and not c.endswith("clustered_pass_rate")]

plot_sweep_summary(
    SWEEP_DF,
    sweep_cols,
    metric_col=pass_rate_cols,
    heatmap_col1="bb_ca_noise_scale",
    heatmap_col2="latents_noise_scale",
    marginal_sweep_cols=["bb_ca_noise_scale", "latents_noise_scale"],
    task_col="dataset_name",
    ylabel="pass_rate",
    higher_is_better=True,
    value_labels=VALUE_LABELS,
    # Comment out cols_to_iterate to marginalize over ckpt_path instead of making separate plots for each checkpoint
    cols_to_iterate=["ckpt_path"],
)
```
